```python
def read_input(infile):
    instructions = []
    with open(infile, 'r') as inf:
        for line in inf.readlines():
            instructions.append(line.strip().split())

    for ins in instructions:
        for i in [1, 2]:
            if ins[i].lstrip('-').isdigit():
                ins[i] = int(ins[i])
    return instructions
```

```python
class Registers(dict):
    @staticmethod
    def __missing__(k):
        return k

def to_int(regs, i):
    try:
        return int(i)
    except ValueError:
        return regs[i]

def setr(regs, i, r, v):
    regs[r] = regs[v]
    return i+1

def sub(regs, i, r, v):
    regs[r] -= regs[v]
    return i+1

def mul(regs, i, r, v):
    regs[r] *= regs[v]
    return i+1

def jnz(regs, i, r, v):
    if regs[r] != 0:
        return i+regs[v]
    else:
        return i+1

def process(instructions):
    cmdmap = {'set': setr, 'sub': sub, 'mul': mul, 'jnz': jnz}

    regs = Registers()
            
    regs['a'] = 0
    regs['b'] = 0
    regs['c'] = 0
    regs['d'] = 0
    regs['e'] = 0
    regs['f'] = 0
    regs['g'] = 0
    regs['h'] = 0

    i = 0
    n_mul = 0

    try:
        while i>=0 and i < len(instructions):
            if instructions[i][0] == 'mul':
                # print(i)
                n_mul += 1
            i = cmdmap[instructions[i][0]](regs, i, *instructions[i][1:])
    except:
        return n_mul
    return n_mul

def process2(instructions):
    cmdmap = {'set': setr, 'sub': sub, 'mul': mul, 'jnz': jnz}

    regs = Registers()
            
    regs['a'] = 1
    regs['b'] = 0
    regs['c'] = 0
    regs['d'] = 0
    regs['e'] = 0
    regs['f'] = 0
    regs['g'] = 0
    regs['h'] = 0

    for i in range(8):
        ins = instructions[i]
        i = cmdmap[ins[0]](regs, i, *ins[1:])

    n = 0
    for v in range(regs['b'], regs['c']+1, 17):
        for i in range(2, v):
            if v % i == 0:
                n += 1
                break
    return n
```

```python
print('*******\nPuzzle1\n*******\n')

print('Puzzle case\n-----------\n')

res = process(read_input('input23.txt'))

print(f'mul is invoked {res} times')

assert res == 6724

print('\n*******\nPuzzle2\n*******\n')

print('Puzzle case\n-----------\n')

res = process2(read_input('input23.txt'))

print(f'The value in register h is {res}')

assert res == 903
```
