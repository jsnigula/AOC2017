```python
import numpy as np
```

```python
def read_input(infile):
    states = {}
    state = {}
    start = ''
    nsteps = 0
    state_name = ''
    with open(infile, 'r') as inf:
        for line in inf.readlines():
            words = line.strip().rstrip('.').split()
            if len(words) == 0:
                if state_name != '':
                    states[state_name] = state
                # Start a new state
                state = {0: [], 1: []}
                state_name = ''
                state_val = -1
            elif words[0] == 'Begin':
                start = words[3]
            elif words[0] == 'Perform':
                nsteps = int(words[5])
            elif words[0] == 'In':
                state_name = words[2].rstrip(':')
            elif words[0] == 'If':
                state_val = int(words[5].rstrip(':'))
            elif words[0] == '-':
                if words[1] == 'Write':
                    state[state_val].append(int(words[4]))
                if words[1] == 'Move':
                    state[state_val].append(1 if words[6] == 'right' else -1)
                if words[1] == 'Continue':
                    state[state_val].append(words[4])

    states[state_name] = state

    return states, nsteps

def run(states, nsteps):

    # print(states)
    ones = np.zeros(nsteps*2+1, dtype=int)

    p = nsteps+1
    s = 'A'

    for i in range(nsteps):
        v, d, n = states[s][ones[p]]
        ones[p] = v
        p += d
        s = n
    return int(np.sum(ones))
```

```python
print('*******\nPuzzle1\n*******\n')

print('Test case\n---------\n')

res = run(*read_input('input25a.txt'))

print(f'Number of ones is {res}')

assert res == 3

print('\nPuzzle case\n-----------\n')

res = run(*read_input('input25.txt'))

print(f'Number of ones is {res}')

assert res == 3732
```
