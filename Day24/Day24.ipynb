```python
from collections import defaultdict
```

```python
def read_input(infile):
    parts = []
    pdict = defaultdict(list)
    with open(infile, 'r') as inf:
        for line in inf.readlines():
            p = tuple([int(v) for v in line.strip().split('/')])
            if p in parts:
                print('Dup', p)
            parts.append(p)
            pdict[p[0]].append(p)
            pdict[p[1]].append(p)
    return parts, pdict
```

```python
def append_bridge(bridge, end, parts, pdict):

    bridges = [bridge]
    # print(bridge)
    for p in pdict[end]:
        # print('Checking', p)
        if p not in bridge:
            # print('Using', p)
            next_end = p[0] if p[0] != end else p[1]
            bridges += append_bridge(bridge + [p], next_end, parts, pdict)
    # print('Returning', bridges)
    return bridges

def build_bridge(parts, pdict):

    max_strength = 0
    max_strength_l = 0

    done = False

    bridges = []
    for p in pdict[0]:
        next_end = p[0] if p[0] != 0 else p[1]
        bridges += append_bridge([p], next_end, parts, pdict)

    max_l = 0
    for b in bridges:
        max_l = max(len(b), max_l)
        
    for b in bridges:
        s = 0
        for p in b:
            s += p[0] + p[1]
        max_strength = max(max_strength, s)
        if len(b) == max_l:
            max_strength_l = max(max_strength_l, s)

    
    return max_strength, max_strength_l
        

build_bridge(*read_input('input24.txt'))
```

```python
print('*******\nPuzzle1\n*******\n')

print('Test case\n---------\n')

t_res, t_res_l = build_bridge(*read_input('input24a.txt'))

print(f'Max strength is {t_res}')

assert t_res == 31

print('\nPuzzle case\n-----------\n')

res, res_l = build_bridge(*read_input('input24.txt'))

print(f'Max strength is {res}')

assert res == 1940

print('\n*******\nPuzzle2\n*******\n')

print('Test case\n---------\n')

print(f'Max strength of longest bridge is {t_res_l}')

assert t_res_l == 19

print('\nPuzzle case\n-----------\n')

print(f'Max strength of longest bridge is {res_l}')

assert res_l == 1928
```
